```python
import numpy as np
import random
```

```python

grid_size = 5


start = (0, 0)
goal = (4, 4)


obstacles = [(1, 1), (2, 2), (3, 1)]


actions = {
    0: (-1, 0),
    1: (1, 0),
    2: (0, -1),
    3: (0, 1)
}


q_table = np.zeros((grid_size, grid_size, 4))


learning_rate = 0.1
discount_factor = 0.9
epsilon = 0.2
episodes = 1000


for episode in range(episodes):

    state = start
    done = False

    while not done:

        row, col = state


        if random.random() < epsilon:
            action = random.randint(0, 3)
        else:
            action = np.argmax(q_table[row, col])


        dr, dc = actions[action]
        new_row = row + dr
        new_col = col + dc


        if not (0 <= new_row < grid_size and
                0 <= new_col < grid_size):
            new_row, new_col = row, col

        next_state = (new_row, new_col)


        if next_state in obstacles:
            reward = -10
            next_state = state
        elif next_state == goal:
            reward = 100
            done = True
        else:
            reward = -1


        old_value = q_table[row, col, action]

        next_max = np.max(
            q_table[next_state[0], next_state[1]]
        )

        new_value = old_value + learning_rate * (
            reward + discount_factor * next_max - old_value
        )

        q_table[row, col, action] = new_value

        state = next_state


state = start
path = [state]

for step in range(30):

    if state == goal:
        break

    row, col = state
    action = np.argmax(q_table[row, col])

    dr, dc = actions[action]
    new_state = (row + dr, col + dc)

    if not (0 <= new_state[0] < grid_size and
            0 <= new_state[1] < grid_size):
        break

    if new_state in obstacles:
        break

    state = new_state
    path.append(state)

print("Learned Path:")
print(path)

print("\nGrid World:")

for i in range(grid_size):
    for j in range(grid_size):

        if (i, j) == start:
            print("S", end="  ")
        elif (i, j) == goal:
            print("G", end="  ")
        elif (i, j) in obstacles:
            print("X", end="  ")
        elif (i, j) in path:
            print("*", end="  ")
        else:
            print(".", end="  ")

    print()
```
